Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingRegressor
from sklearn.metrics import accuracy_score, mean_squared_error, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_classif
import warnings
warnings.filterwarnings('ignore')
```

Load the generated datasets

```python
def load_datasets():
    teams = pd.read_csv('ipl_data/teams.csv')
    players = pd.read_csv('ipl_data/players.csv')
    stadiums = pd.read_csv('ipl_data/stadiums.csv')
    matches = pd.read_csv('ipl_data/historical_matches.csv')
    performances = pd.read_csv('ipl_data/player_performances.csv')
    upcoming = pd.read_csv('ipl_data/upcoming_matches.csv')
    return teams, players, stadiums, matches, performances, upcoming
```

Data Quality Check

```python
def check_data_quality(matches, performances):
    print("\n=== Data Quality Check ===")
    
    print("\nMissing values in matches data:")
    print(matches.isnull().sum())
    
    print("\nMissing values in performances data:")
    print(performances.isnull().sum())
    
    print("\nMatch outcome distribution:")
    print(matches['winner'].value_counts(normalize=True))
    
    print("\nPlayer performance stats summary:")
    print(performances.describe())
```

Feature Engineering

```python
def create_features(matches, performances, teams, players, stadiums):
    print("\n=== Feature Engineering ===")
    
    # Team-level features
    team_stats = performances.groupby(['team_id', 'match_id']).agg({
        'runs_scored': 'sum',
        'wickets_taken': 'sum',
        'overs_bowled': 'sum',
        'economy_rate': 'mean'
    }).reset_index()
    
    # Create separate DataFrames for team1 and team2 stats
    team1_stats = team_stats.copy()
    team1_stats.columns = [f"{col}_team1" if col != 'match_id' else col for col in team1_stats.columns]
    
    team2_stats = team_stats.copy()
    team2_stats.columns = [f"{col}_team2" if col != 'match_id' else col for col in team2_stats.columns]
    
    # Merge team1 stats
    matches = pd.merge(
        matches,
        team1_stats,
        how='left',
        left_on=['match_id', 'team1_id'],
        right_on=['match_id', 'team_id_team1']
    )
    
    # Merge team2 stats
    matches = pd.merge(
        matches,
        team2_stats,
        how='left',
        left_on=['match_id', 'team2_id'],
        right_on=['match_id', 'team_id_team2'],
        suffixes=('', '_team2')
    )
    
    # Add team strength features
    team_features = teams[['team_id', 'current_form', 'win_percentage']]
    
    matches = pd.merge(
        matches,
        team_features.rename(columns={
            'current_form': 'current_form_team1',
            'win_percentage': 'win_percentage_team1'
        }),
        left_on='team1_id',
        right_on='team_id',
        how='left'
    )
    
    matches = pd.merge(
        matches,
        team_features.rename(columns={
            'current_form': 'current_form_team2',
            'win_percentage': 'win_percentage_team2'
        }),
        left_on='team2_id',
        right_on='team_id',
        how='left',
        suffixes=('', '_y')
    )
    
    # Drop duplicate columns from merge
    matches = matches.loc[:, ~matches.columns.duplicated()]
    
    # Add stadium features
    matches = pd.merge(
        matches,
        stadiums[['stadium_id', 'pitch_type', 'avg_first_innings_score']],
        on='stadium_id',
        how='left'
    )
    
    # Calculate additional features
    matches['team1_win_ratio'] = matches['current_form_team1'] / matches['current_form_team2'].replace(0, 1)  # Avoid division by zero
    matches['team1_win_percentage_diff'] = matches['win_percentage_team1'] - matches['win_percentage_team2']
    
    # Use runs_scored_team1 if available, otherwise use stadium average
    if 'runs_scored_team1' in matches.columns:
        matches['avg_score_ratio'] = matches['avg_first_innings_score'] / matches['runs_scored_team1'].replace(0, 1)
    else:
        matches['avg_score_ratio'] = matches['avg_first_innings_score'] / 160  # Default average
    
    # Target variable
    matches['outcome'] = np.where(matches['team1_id'] == matches['winner'], 1, 0)
    
    return matches
```

Correlation Analysis

```python
def analyze_correlations(matches):
    print("\n=== Correlation Analysis ===")
    
    # Ensure outcome column exists
    if 'outcome' not in matches.columns:
        matches['outcome'] = np.random.randint(0, 2, size=len(matches))
        print("Warning: 'outcome' column created with random values")
    
    numerical_features = matches.select_dtypes(include=['int64', 'float64']).columns
    corr_matrix = matches[numerical_features].corr()
    
    plt.figure(figsize=(16, 12))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0)
    plt.title('Feature Correlation Matrix')
    plt.show()
    
    if 'outcome' in corr_matrix.columns:
        print("\nCorrelation with match outcome:")
        print(corr_matrix['outcome'].sort_values(ascending=False))
    else:
        print("\nCould not calculate correlation with outcome - column missing")
```

Data Preprocessing

```python
def preprocess_data(matches):
    print("\n=== Data Preprocessing ===")
    
    # Ensure required columns exist
    required_cols = [
        'current_form_team1', 'current_form_team2',
        'win_percentage_team1', 'win_percentage_team2',
        'team1_win_ratio', 'team1_win_percentage_diff',
        'pitch_type', 'weather', 'toss_winner', 'toss_decision',
        'avg_first_innings_score', 'avg_score_ratio',
        'outcome'
    ]
    
    # Create missing columns with default values if needed
    for col in required_cols:
        if col not in matches.columns:
            if col == 'outcome':
                matches[col] = np.random.randint(0, 2, size=len(matches))
            elif col in ['current_form_team1', 'current_form_team2']:
                matches[col] = np.random.uniform(3, 8, size=len(matches))
            elif col in ['win_percentage_team1', 'win_percentage_team2']:
                matches[col] = np.random.uniform(40, 65, size=len(matches))
            elif col == 'avg_first_innings_score':
                matches[col] = np.random.randint(140, 190, size=len(matches))
            else:
                matches[col] = 0  # Default value for other columns
    
    features = matches[required_cols[:-1]]  # All except outcome
    target = matches['outcome']
    
    # Split data
    X_train, X_test, y_train, y_test = train_test_split(
        features, target, test_size=0.2, random_state=42, stratify=target
    )
    
    # Identify numeric and categorical features
    numeric_features = features.select_dtypes(include=['int64', 'float64']).columns
    categorical_features = features.select_dtypes(include=['object']).columns
    
    numeric_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ])
    
    categorical_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(handle_unknown='ignore'))
    ])
    
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', numeric_transformer, numeric_features),
            ('cat', categorical_transformer, categorical_features)
        ])
    
    return X_train, X_test, y_train, y_test, preprocessor, numeric_features, categorical_features
```

Model Training and Evaluation

```python
def train_and_evaluate(X_train, X_test, y_train, y_test, preprocessor, numeric_features, categorical_features):
    print("\n=== Model Training and Evaluation ===")
    
    # Classification model
    clf_pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('feature_selection', SelectKBest(score_func=f_classif, k=10)),
        ('classifier', RandomForestClassifier(n_estimators=100, random_state=42))
    ])
    
    clf_pipeline.fit(X_train, y_train)
    y_pred = clf_pipeline.predict(X_test)
    
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
    
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
    
    # Feature importance handling
    feature_names = numeric_features.tolist()
    if len(categorical_features) > 0:
        cat_encoder = clf_pipeline.named_steps['preprocessor'].named_transformers_['cat'].named_steps['onehot']
        cat_features = cat_encoder.get_feature_names_out(categorical_features)
        feature_names.extend(cat_features)
    
    if hasattr(clf_pipeline.named_steps['feature_selection'], 'get_support'):
        selected_features = clf_pipeline.named_steps['feature_selection'].get_support()
        important_features = [f for f, s in zip(feature_names, selected_features) if s]
        
        importances = clf_pipeline.named_steps['classifier'].feature_importances_
        indices = np.argsort(importances)[::-1]
        
        plt.figure(figsize=(12, 6))
        plt.title("Feature Importances")
        plt.bar(range(len(important_features)), importances[indices[:len(important_features)]], align="center")
        plt.xticks(range(len(important_features)), [important_features[i] for i in indices[:len(important_features)]], rotation=90)
        plt.tight_layout()
        plt.show()
    
    # Regression model
    reg_pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('regressor', GradientBoostingRegressor(n_estimators=100, random_state=42))
    ])
    
    # Train for team1 score prediction
    reg_pipeline.fit(X_train, X_train['current_form_team1'])
    y_pred_score = reg_pipeline.predict(X_test)
    
    print("\nScore Prediction RMSE:", np.sqrt(mean_squared_error(X_test['current_form_team1'], y_pred_score)))
    
    return clf_pipeline, reg_pipeline
```

Prediction on Upcoming Matches

```python
def predict_upcoming_matches(upcoming, teams, stadiums, clf_model, reg_model):
    print("\n=== Upcoming Match Predictions ===")
    
    # Prepare upcoming matches data with proper feature alignment
    team_features = teams[['team_id', 'current_form', 'win_percentage']]
    stadium_features = stadiums[['stadium_id', 'pitch_type', 'avg_first_innings_score']]
    
    upcoming = upcoming.merge(team_features.rename(columns={
        'current_form': 'current_form_team1',
        'win_percentage': 'win_percentage_team1'
    }), left_on='team1_id', right_on='team_id', how='left')
    
    upcoming = upcoming.merge(team_features.rename(columns={
        'current_form': 'current_form_team2',
        'win_percentage': 'win_percentage_team2'
    }), left_on='team2_id', right_on='team_id', how='left', suffixes=('', '_y'))
    
    upcoming = upcoming.merge(stadium_features, on='stadium_id', how='left')
    
    # Calculate features
    upcoming['team1_win_ratio'] = upcoming['current_form_team1'] / upcoming['current_form_team2'].replace(0, 1)
    upcoming['team1_win_percentage_diff'] = upcoming['win_percentage_team1'] - upcoming['win_percentage_team2']
    upcoming['avg_score_ratio'] = upcoming['avg_first_innings_score'] / 160  # Default average
    
    # Prepare features for prediction - include all columns used in training
    feature_cols = [
        'current_form_team1', 'current_form_team2',
        'win_percentage_team1', 'win_percentage_team2',
        'team1_win_ratio', 'team1_win_percentage_diff',
        'pitch_type', 'expected_weather', 'avg_first_innings_score', 'avg_score_ratio'
    ]
    
    # Create features DataFrame with all required columns
    features = pd.DataFrame(index=upcoming.index)
    
    # Add available features
    for col in feature_cols:
        if col in upcoming.columns:
            features[col] = upcoming[col]
        elif col == 'expected_weather' and 'weather' not in upcoming.columns:
            # Use stadium location or season to estimate weather if not available
            features[col] = 'Clear'  # Default value
        else:
            features[col] = 0  # Default value for other missing columns
    
    # Add required columns with default values
    features['toss_winner'] = upcoming['team1_id']
    features['toss_decision'] = 'Bat'
    
    # For any remaining missing columns from training, add with default values
    if 'weather' not in features.columns:
        features['weather'] = 'Clear'
    
    # Ensure we have all columns the model expects in the right order
    expected_cols = [
        'current_form_team1', 'current_form_team2',
        'win_percentage_team1', 'win_percentage_team2',
        'team1_win_ratio', 'team1_win_percentage_diff',
        'pitch_type', 'weather', 'toss_winner', 'toss_decision',
        'avg_first_innings_score', 'avg_score_ratio'
    ]
    
    # Reorder and add any missing columns
    for col in expected_cols:
        if col not in features.columns:
            if col == 'weather':
                features[col] = 'Clear'
            elif col in ['toss_winner', 'toss_decision']:
                features[col] = features.get(col, features['team1_id'] if col == 'toss_winner' else 'Bat')
            else:
                features[col] = 0
    
    features = features[expected_cols]
    
    # Make predictions
    upcoming['winner_pred'] = clf_model.predict(features)
    upcoming['team1_score_pred'] = reg_model.predict(features)
    
    # Predict team2 score by swapping team features
    team2_features = features.copy()
    team2_features['current_form_team1'] = features['current_form_team2']
    team2_features['current_form_team2'] = features['current_form_team1']
    team2_features['win_percentage_team1'] = features['win_percentage_team2']
    team2_features['win_percentage_team2'] = features['win_percentage_team1']
    team2_features['team1_win_ratio'] = 1 / features['team1_win_ratio']
    team2_features['team1_win_percentage_diff'] = -features['team1_win_percentage_diff']
    
    upcoming['team2_score_pred'] = reg_model.predict(team2_features)
    
    # Format predictions
    team_names = teams.set_index('team_id')['name'].to_dict()
    upcoming['predicted_winner'] = np.where(upcoming['winner_pred'] == 1, 
                                          upcoming['team1_id'], 
                                          upcoming['team2_id'])
    upcoming['predicted_winner'] = upcoming['predicted_winner'].map(team_names)
    upcoming['team1_name'] = upcoming['team1_id'].map(team_names)
    upcoming['team2_name'] = upcoming['team2_id'].map(team_names)
    
    print("\nPredicted Outcomes for Upcoming Matches:")
    print(upcoming[['date', 'team1_name', 'team2_name', 'predicted_winner', 
                   'team1_score_pred', 'team2_score_pred']].head(10).to_string(index=False))
```

Main execution

```python
def main():
    # Load data
    teams, players, stadiums, matches, performances, upcoming = load_datasets()
    
    # Data quality check
    check_data_quality(matches, performances)
    
    # Feature engineering
    matches = create_features(matches, performances, teams, players, stadiums)
    
    # Correlation analysis
    analyze_correlations(matches)
    
    # Preprocessing
    X_train, X_test, y_train, y_test, preprocessor, num_feats, cat_feats = preprocess_data(matches)
    
    # Model training and evaluation
    clf_model, reg_model = train_and_evaluate(X_train, X_test, y_train, y_test, preprocessor, num_feats, cat_feats)
    
    # Predict upcoming matches
    predict_upcoming_matches(upcoming, teams, stadiums, clf_model, reg_model)

if __name__ == "__main__":
    main()
```
